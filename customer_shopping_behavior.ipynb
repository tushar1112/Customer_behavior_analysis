```python
import pandas as pd
```

```python
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df['Review Rating'] = df['Review Rating'].fillna(df.groupby('Category')['Review Rating'].transform('median'))

df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))   # same as above using lamda
```

```python
df.isnull().sum()
```

#### making changes to column names into snake case, i.e all lower with underscore for space

```python
df.columns = df.columns.str.lower().str.replace(' ','_')
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

#### Creating new columns as feature engineering

```python
# grouping age based on age group

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
bins = [0, 18, 35, 50, 100]
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)
```

```python
df['frequency_of_purchases'].unique()
```

```python
# creating column purchase_frequency_day -> converting string based values like 'weekly', 'monthly' to 7, 30

frequency_mapping = {
    'Fortnightly' : 14, 
    'Weekly' : 7, 
    'Annually' : 365, 
    'Quarterly' : 120, 
    'Bi-Weekly' : 14,
    'Monthly' : 30, 
    'Every 3 Months' : 120
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df.head()
```

```python
# checking if column 'discount_applied' and 'promo_code_used' is identical or not.

(df['promo_code_used'] == df['discount_applied']).all()
```

```python
# it gives true meaning all values match for both the column. Hence redundant data and drop one

df = df.drop('promo_code_used', axis=1)
```
